```python
df1 = df[['gender','ssc_p','hsc_p','degree_p','workex','etest_p','mba_p','status','salary']]
df1
```

DEGREE% VS PLACEMENT

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.boxplot(x = 'status',y='degree_p',data = df1)
plt.show()
```

MBA% VS PLACEMENT

```python
sns.boxplot(x = 'status',y = 'mba_p',data = df1)
plt.show()
```

GENDER VS PLACEMENT

```python
sns.countplot(x = 'gender',hue = 'status',data = df1)
plt.show()
```

```python
gender_placement = pd.crosstab(df1['gender'],df1['status'],normalize = 'index')*100
gender_placement
```

WORK EXPERIENCE VS PLACEMENT

```python
sns.countplot(x = 'workex', hue ='status',data = df1)
plt.show()
```

ENTRANCE TEST VS PLACEMENT

```python
sns.boxplot(x = 'status',y = 'etest_p',data = df1)
plt.show()
```

placed students

```python
placed_df = df[df['status'] == 'Placed']
placed_df
```

degree% vs salary

```python
sns.scatterplot(x = 'degree_p',y = 'salary',data = placed_df)
plt.figure(figsize = (10,10))
plt.show()
```

MBA% VS SALARY

```python
sns.scatterplot(x = 'mba_p',y = 'salary',data = placed_df)
plt.show()
```

CORRELATION

```python
sns.heatmap(df1.corr(numeric_only = True),annot = True)
plt.show()
```

PLACEMENT PERCENTAGE

```python
placement_percentage = df['status'].value_counts(normalize = True)*100
placement_percentage
```

Comparing Degree and MBA together for Placements
